# Concatenación de los Conteos Censales 2009-2024

Lee los CSV de `Datos/datos_ine/` (6 elecciones federales × 32 entidades), conserva las variables seleccionadas y genera `Datos/datos_procesados/concatenacion/concatenacion_2009-2024.csv`.

El procesamiento es **por archivo** (se agrega cada entidad al CSV de salida) porque el total pesa ~3.7 GB y no cabe cómodamente en memoria.

```python
from pathlib import Path
import pandas as pd
```

```python
# Raíz del proyecto: se sube desde la carpeta actual hasta encontrar la carpeta Datos/
_cwd = Path.cwd().resolve()
RAIZ = next(p for p in [_cwd, *_cwd.parents] if (p / "Datos").is_dir())
DIR_ENTRADA = RAIZ / "Datos" / "datos_ine"
ARCHIVO_SALIDA = RAIZ / "Datos" / "datos_procesados" / "concatenacion" / "concatenacion_2009-2024.csv"

VARIABLES = [
    "AELEC", "FELECCION", "EDONOM", "MPIONOM",
    "SECCION", "DEF", "TIPOSEC", "SEXO", "EDAD", "LN", "SV", "NV", "NS",
]

print("Raíz del proyecto:", RAIZ)
print("Entrada existe:", DIR_ENTRADA.exists())
```

```python
# Un CSV por entidad y año; los .txt de descripción de campos no coinciden con *.csv
archivos = sorted(DIR_ENTRADA.glob("*/*.csv"))
print(f"{len(archivos)} archivos encontrados (esperados: 192 = 6 años × 32 entidades)")
archivos[:3]
```

```python
ARCHIVO_SALIDA.parent.mkdir(parents=True, exist_ok=True)
if ARCHIVO_SALIDA.exists():
    ARCHIVO_SALIDA.unlink()  # se regenera desde cero

resumen = []
for i, ruta in enumerate(archivos, start=1):
    df = pd.read_csv(
        ruta,
        usecols=VARIABLES,
        encoding="utf-8-sig",  # los archivos traen BOM
        dtype={"FELECCION": "string", "EDONOM": "string", "MPIONOM": "string", "TIPOSEC": "string"},
    )[VARIABLES]
    df.to_csv(
        ARCHIVO_SALIDA,
        mode="a",
        header=not ARCHIVO_SALIDA.exists(),
        index=False,
        encoding="utf-8-sig",
    )
    resumen.append({"archivo": ruta.name, "filas": len(df)})
    if i % 16 == 0 or i == len(archivos):
        print(f"[{i}/{len(archivos)}] {ruta.name}")

resumen = pd.DataFrame(resumen)
print(f"\nFilas totales escritas: {resumen['filas'].sum():,}")
```

```python
# Verificación: filas por año a partir del nombre de archivo
resumen["anio"] = resumen["archivo"].str.extract(r"conteoscensales(\d{4})")
print(resumen.groupby("anio")["filas"].sum())
print(f"\nTamaño del CSV: {ARCHIVO_SALIDA.stat().st_size / 1e9:.2f} GB")
```
